```python
import sys
import asyncio

# Fix for Windows asyncio subprocess NotImplementedError in Jupyter
if sys.platform == 'win32':
    asyncio.set_event_loop_policy(asyncio.WindowsProactorEventLoopPolicy())

import re
import pandas as pd
from playwright.async_api import async_playwright

# Start Playwright asynchronously
playwright_obj = await async_playwright().start()

# Launch a Chromium browser window
browser = await playwright_obj.chromium.launch(headless=False)
context = await browser.new_context(
    user_agent="Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36",
    viewport={"width": 1280, "height": 800}
)
page = context.new_page()

print("Browser initialized successfully without errors!")
```
